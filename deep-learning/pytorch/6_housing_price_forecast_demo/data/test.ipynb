```python
import pandas as pd

# 创建数据字典
data = {
    '姓名': ['张三', '李四', '王五'],
    '年龄': [28, 34, 29],
    '城市': ['北京', '上海', '广州']
}

# 实例化 DataFrame
df = pd.DataFrame(data)

print("--- 基础 DataFrame ---")
print(df)
```

```python
# 查看前 2 行数据
print("--- head() 预览 ---")
print(df.head(2))

print("\n--- info() 结构信息 ---")
df.info()

print("\n--- describe() 统计学描述 ---")
print(df.describe())
```

```python
import numpy as np
from scipy.sparse import csr_matrix

# 1. 模拟一个稀疏矩阵的底层存储
# 假设是一个 3x3 的矩阵，大部分是 0
data = np.array([1, 1, 1])
row = np.array([0, 1, 2])
col = np.array([0, 1, 0])
sparse_matrix = csr_matrix((data, (row, col)), shape=(3, 3))

print("--- 转换前（稀疏矩阵） ---")
print(sparse_matrix)

# 2. 使用 .toarray() 还原
dense_array = sparse_matrix.toarray()

print("\n--- 转换后（密集型二维数组） ---")
print(dense_array)
```
